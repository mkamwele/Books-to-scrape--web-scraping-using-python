```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url="https://books.toscrape.com/catalogue/category/books_1/index.html"
```

```python

response=requests.get(url)
```

```python
response.status_code
```

```python
# Parse the HTML content
soup=BeautifulSoup(response.content)
```

```python
#Find the side_categories section
side_categories=soup.find("div",class_="side_categories")
```

```python
#select the links inside the nested <ul>
categories = side_categories.find("ul").find_all("li")
```

```python
for li in categories:
    print(li.a.get_text(strip=True))
```

```python
category_links=side_categories.select("ul ul li a")
for link in category_links:
     print(link.get_text(strip=True))
```

```python
travel_link = None

for link in side_categories.find_all("a"):
    if link.get_text(strip=True) == "Travel":
        travel_link = link
        break
```

```python
print(travel_link)
```

```python
travel_href = travel_link.get("href")

print(travel_href)
```

```python
#Then convert it to the actual URL
from urllib.parse import urljoin

travel_url = urljoin(url, travel_href)

print(travel_url)
```

```python
response = requests.get(travel_url)

travel_soup = BeautifulSoup(response.text, "html.parser")
```

```python
#Find all the individual books
books = travel_soup.find_all("article", class_="product_pod")
```

```python
print(len(books))
```

```python
print(books[0])
```

```python
#Extract the title
title=books[0].h3.a.get("title")
print(title)
```

```python
#. Extract the price
price=books[0].find(
    "p",
    class_="price_color"
).get_text(strip=True)
print(price)
```

```python
#Extract availability
availability=books[0].find(
    "p",
    class_="instock availability"
).get_text(strip=True)
print(availability)
```

```python
#put it into a loop
books_data = []
for book in books:
    title=book.h3.a.get("title")
    price = book.find(
        "p",
        class_="price_color"
    ).get_text(strip=True)

    availability = book.find(
        "p",
        class_="instock availability"
    ).get_text(strip=True)
    
    books_data.append({
        "Title": title,
        "Price": price,
        "Availability": availability
    })
```

```python
travel_books_df = pd.DataFrame(books_data)
```

```python
"""
scrape all books in each category and save it in a data frame:
Finds every book category.
Gets each category's URL.
Visits each category page.
Scrapes every book on that page.
Follows the Next button to get additional pages.
Stores everything in one pandas DataFrame.
"""
```

```python
#find the category <a> tags
outer_ul = side_categories.find("ul")
category_list = outer_ul.find("ul")
category_links=category_list.find_all("a")
```

```python
for link in category_links:
    print(link.get_text(strip=True), link.get("href"))
```

```python
#Create an empty list for all your data
all_books = []
```

```python
for category_link in category_links:
    category=category_link.get_text(strip=True)
    category_href=category_link.get("href")
    category_url = urljoin(url, category_href)
    response=requests.get(category_url)
    category_soup=BeautifulSoup(response.text,"html.parser")
    books=category_soup.find_all("article", class_="product_pod")
    for book in books:
        title=book.h3.a.get("title")
        price=float(book.find("p",class_="price_color").get_text(strip=True).replace("£","").replace("Â",""))
        availability=book.find("p",class_="instock availability").get_text(strip=True)
        all_books.append({
            "Category":category,
            "Title":title,
            "Price":price,
            "Availability":availability
        })
```

```python
# Find the nested <ul> containing the categories
category_list=side_categories.find("ul").find("ul")
# Find all category links
category_links=category_list.find_all("a")
```

```python
all_books=[]
for category_link in category_links:

    category = category_link.get_text(strip=True)
    category_href = category_link.get("href")
    category_url = urljoin(url, category_href)

    current_url = category_url

    while current_url:

        print("Scraping:", current_url)

        response = requests.get(current_url, timeout=10)

        category_soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        books = category_soup.find_all(
            "article",
            class_="product_pod"
        )

        for book in books:

            title = book.h3.a.get("title")

            price = float(
                book.find(
                    "p",
                    class_="price_color"
                )
                .get_text(strip=True)
                .replace("£", "")
                .replace("Â", "")
            )

            availability = book.find(
                "p",
                class_="instock availability"
            ).get_text(strip=True)

            all_books.append({
                "Category": category,
                "Title": title,
                "Price": price,
                "Availability": availability
            })

        # Look for the "next" button
        next_link = category_soup.find(
            "li",
            class_="next"
        )

        if next_link:

            next_href = next_link.a.get("href")

            current_url = urljoin(
                current_url,
                next_href
            )

        else:

            current_url = None
```

```python
books_df = pd.DataFrame(all_books)

books_df.shape
```
